```python
import pandas as pd
import numpy as np
import tensorflow as tf
```

```python
# Make sure 'california_housing_train.csv' is uploaded to your Colab environment or replace 'california_housing_train.csv' with the correct path to your Excel file.
dataset=pd.read_csv('california_housing_train.csv')
X=dataset.iloc[:,:-1].values
Y=dataset.iloc[:,-1].values
print(X)
print(Y)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=0)
```

```python
ann=tf.keras.models.Sequential()
```

```python
ann.add(tf.keras.layers.Dense(units=3,activation='relu'))
ann.add(tf.keras.layers.Dense(units=3,activation='relu'))
ann.add(tf.keras.layers.Dense(units=3,activation='relu'))
ann.add(tf.keras.layers.Dense(units=1))
```

```python
ann.compile(optimizer='adam', loss='mean_squared_error')
```

```python
ann.fit(X_train,Y_train,batch_size=64,epochs=100)
```
